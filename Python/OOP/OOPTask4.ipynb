# Task 1.
- Calculate Factorial using recursion

```python
def factorial(num):
    if num == 0:
        return 0
    if num == 1:
        return 1
    return num * factorial(num -1 )

factorial(5)
```

```python
class Task1:
    """
    Task1 class handles calculating the factorial of a number using recursion.

    Attributes:
        number (int): The number whose factorial will be calculated.
    """

    def __init__(self, number):
        """
        Initializes the Task1 object with the number to calculate.

        Args:
            number (int): The number whose factorial will be calculated.
        """
        self.number = number

    def calculate_factorial(self, number=None):
        """
        Calculates the factorial of a number using recursion.

        Args:
            number (int, optional): The number to calculate the factorial for.
                Defaults to self.number when not provided.

        Returns:
            int: The factorial of the number.
        """
        if number is None:
            number = self.number

        # base case
        if number == 0 or number == 1:
            return 1
        # recursive case
        return number * self.calculate_factorial(number - 1)


#### test your class
if __name__ == "__main__":
    task1 = Task1(5)
    result = task1.calculate_factorial()
    print(f"Factorial of 5 is: {result}")
```

# Task 2.
- Write a function that takes a number and return True if this number is prime and False if not.

```python
def isPrime(num):
    for i in range(2, num):
        if num % i == 0:
            return False
    return True

isPrime(20)
```

```python
class Task2:
    """
    Task2 class checks whether a given number is a prime number.

    Attributes:
        number (int): The number to check for primality.
    """

    def __init__(self, number):
        """
        Initializes the Task2 object with the number to check.

        Args:
            number (int): The number to check for primality.
        """
        self.number = number

    def is_prime(self):
        """
        Determines whether the number is prime.

        Returns:
            bool: True if the number is prime, otherwise False.
        """
        if self.number < 2:
            return False
        for i in range(2, self.number):
            if self.number % i == 0:
                return False
        return True


#### test your class
if __name__ == "__main__":
    task2 = Task2(20)
    print(f"Is 20 prime? {task2.is_prime()}")

    task2_b = Task2(17)
    print(f"Is 17 prime? {task2_b.is_prime()}")
```

# Task 3.
- Write a function that takes two numbers and return a list of common dividors

```python
def dividors(num1, num2):
    mn = min(num1, num2)
    common_dividors = []
    for i in range(1, mn+1):
        if num1 % i == 0 and num2 % i ==0:
            common_dividors.append(i)
    return common_dividors

dividors(20, 10)
```

```python
class Task3:
    """
    Task3 class finds the common divisors of two numbers.

    Attributes:
        num1 (int): The first number.
        num2 (int): The second number.
    """

    def __init__(self, num1, num2):
        """
        Initializes the Task3 object with the two numbers.

        Args:
            num1 (int): The first number.
            num2 (int): The second number.
        """
        self.num1 = num1
        self.num2 = num2

    def get_common_dividors(self):
        """
        Finds all common divisors of num1 and num2.

        Returns:
            list: A list of integers that divide both numbers evenly.
        """
        smaller = min(self.num1, self.num2)
        common_dividors = []
        for i in range(1, smaller + 1):
            if self.num1 % i == 0 and self.num2 % i == 0:
                common_dividors.append(i)
        return common_dividors


#### test your class
if __name__ == "__main__":
    task3 = Task3(20, 10)
    print(f"Common divisors of 20 and 10: {task3.get_common_dividors()}")
```

```python
class Task4:
    """
    Task4 class finds the longest common substring between two strings.

    Attributes:
        str1 (str): The first string.
        str2 (str): The second string.
    """

    def __init__(self, str1, str2):
        """
        Initializes the Task4 object with the two strings.

        Args:
            str1 (str): The first string.
            str2 (str): The second string.
        """
        self.str1 = str1
        self.str2 = str2

    def find_longest_common_substring(self):
        """
        Finds the longest common substring between str1 and str2.

        Returns:
            str: The longest substring that appears in both strings.
        """
        longest_str = ""
        for i in range(len(self.str1)):
            for j in range(i, len(self.str1)):
                sub_str = self.str1[i:j + 1]
                if sub_str in self.str2:
                    if len(sub_str) > len(longest_str):
                        longest_str = sub_str
        return longest_str


#### test your class
if __name__ == "__main__":
    str1 = "Welcome to Machine Learning Diploma"
    str2 = "I am studying a Machine Learning Course"
    task4 = Task4(str1, str2)
    print(f"Longest common substring: {task4.find_longest_common_substring()}")
```

```python
class Task5:
    """
    Task5 class reverses the order of words in a given string.

    Attributes:
        text (str): The string whose words will be reversed.
    """

    def __init__(self, text):
        """
        Initializes the Task5 object with the text to reverse.

        Args:
            text (str): The string whose words will be reversed.
        """
        self.text = text

    def reverse_words(self):
        """
        Reverses the order of words in the stored text.

        Returns:
            str: The text with word order reversed.
        """
        words = self.text.split(" ")
        reversed_words = words[::-1]
        return " ".join(reversed_words)


#### test your class
if __name__ == "__main__":
    task5 = Task5("Hello world")
    print(f"Reversed: {task5.reverse_words()}")
```

```python
class Task7:
    """
    Task7 class counts the frequency of each unique word in a list of words.

    Attributes:
        words (list): The list of words to analyze.
    """

    def __init__(self, words):
        """
        Initializes the Task7 object with the list of words.

        Args:
            words (list): The list of words to analyze.
        """
        self.words = words

    def count_frequencies(self):
        """
        Counts how many times each unique word appears in the list.

        Returns:
            dict: A dictionary mapping each word to its frequency.
        """
        unique_words = set(self.words)
        counts = {word: self.words.count(word) for word in unique_words}
        return counts


#### test your class
if __name__ == "__main__":
    words = ["Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"]
    task7 = Task7(words)
    print(f"Word frequencies: {task7.count_frequencies()}")
```

# Task 8.
- Write a Python program to find all the unique words and count the frequency of occurrence from a text file and save the counts in a new file
- This is the same as Task 7 but this time read words from a file
- Hint: use replace function and split function

**First Step: read the file**

```python

input_file = open("input_file.txt", "r")
text = input_file.read()
input_file.close()
```

**Second step: remove '\n' then convert the file text into a list of words**

```python
text = text.replace('\n', ' ')
words = text.split(' ')
```

**Third Step: get unique words, create a dictionary, then count unique words**

```python
unique_words = set(words)
word_counts = dict()

for word in unique_words:
    word_counts[word] = words.count(word)
```

**Save the word-count pair into a text file**

```python
output_file = open("output_file.txt", "w")
for word in word_counts:
    output_file.write("{}: {}\n".format(word, word_counts[word]))
output_file.close()
```

```python
class Task8:
    """
    Task8 class reads words from a text file, counts the frequency of each
    unique word, and saves the counts to an output text file.

    Attributes:
        input_path (str): Path to the input text file.
        output_path (str): Path to the output text file.
        word_counts (dict): The most recently computed word frequency counts.
    """

    def __init__(self, input_path, output_path):
        """
        Initializes the Task8 object with input and output file paths.

        Args:
            input_path (str): Path to the input text file.
            output_path (str): Path to the output text file.
        """
        self.input_path = input_path
        self.output_path = output_path
        self.word_counts = {}

    def read_words(self):
        """
        Reads the input file and splits its content into a list of words.

        Returns:
            list: The list of words found in the input file.
        """
        with open(self.input_path, "r") as input_file:
            text = input_file.read()
        text = text.replace("\n", " ")
        words = text.split(" ")
        return words

    def count_word_frequencies(self):
        """
        Counts the frequency of each unique word in the input file.

        Returns:
            dict: A dictionary mapping each word to its frequency.
        """
        words = self.read_words()
        unique_words = set(words)
        self.word_counts = {word: words.count(word) for word in unique_words}
        return self.word_counts

    def save_counts(self):
        """
        Saves the word frequency counts to the output file.

        Returns:
            None
        """
        with open(self.output_path, "w") as output_file:
            for word, count in self.word_counts.items():
                output_file.write("{}: {}\n".format(word, count))


#### test your class
if __name__ == "__main__":
    task8 = Task8("input_file.txt", "output_file.txt")
    counts = task8.count_word_frequencies()
    task8.save_counts()
    print(f"Word counts saved to output_file.txt: {counts}")
```

```python
import random

# Predefined responses
responses = {
    "hello": ["Hello!", "Hi there!", "Greetings!"],
    "how are you": ["I'm doing well, thank you!", "I'm fine, how about you?"],
    "goodbye": ["Goodbye!", "See you later!", "Farewell!"],
    "default": ["I'm sorry, I didn't understand.", "Could you please rephrase that?"]
}

# Function to get a response based on user input
def get_response(user_input):
    for key in responses:
        if key in user_input:
            return random.choice(responses[key])
    
    return random.choice(responses["default"])

# Chatbot function
def chatbot():
    print("Chatbot: Hi! How can I assist you today?")
    
    while True:
        user_input = input("User: ").lower()
        response = get_response(user_input)
        print("Chatbot:", response)
        
        if user_input == "goodbye":
            break

# Run the chatbot
chatbot()
```

```python
import random


class Task9:
    """
    Task9 class implements a basic chatbot that responds to user input
    using a set of predefined responses.

    Attributes:
        responses (dict): A dictionary mapping keywords to possible replies.
    """

    def __init__(self, responses=None):
        """
        Initializes the Task9 object with predefined responses.

        Args:
            responses (dict, optional): A dictionary of keyword-response pairs.
                If None, a default set of responses is used.
        """
        self.responses = responses if responses is not None else {
            "hello": ["Hello!", "Hi there!", "Greetings!"],
            "how are you": ["I'm doing well, thank you!", "I'm fine, how about you?"],
            "goodbye": ["Goodbye!", "See you later!", "Farewell!"],
            "default": ["I'm sorry, I didn't understand.", "Could you please rephrase that?"]
        }

    def get_response(self, user_input):
        """
        Finds an appropriate response based on the user's input.

        Args:
            user_input (str): The message typed by the user.

        Returns:
            str: A randomly chosen response matching the input, or a default reply.
        """
        for key in self.responses:
            if key in user_input:
                return random.choice(self.responses[key])
        return random.choice(self.responses["default"])

    def start_chat(self):
        """
        Runs an interactive chat loop until the user says 'goodbye'.

        Returns:
            None
        """
        print("Chatbot: Hi! How can I assist you today?")
        while True:
            user_input = input("User: ").lower()
            response = self.get_response(user_input)
            print("Chatbot:", response)
            if user_input == "goodbye":
                break


#### test your class
if __name__ == "__main__":
    task9 = Task9()
    print(task9.get_response("hello"))
    print(task9.get_response("how are you"))
    print(task9.get_response("goodbye"))
    # To chat interactively, uncomment the line below:
    # task9.start_chat()
```

# Task 10: Person class and child Student class 
Create a Python class Person with attributes: name and age.
- Create a display() method that displays the name and age of an object created via the Person class.
- Create a child class Student which inherits from the Person class and which also has an extra attribute called school.
- Create a method displayStudent() that displays the name, age and section of an object created via the Student class.
- Create a student object via an instantiation on the Student class and then test the displayStudent method

```python
class Person:
    """
    Person class represents a person with a name and an age.

    Attributes:
        name (str): The person's name.
        age (int): The person's age.
    """

    def __init__(self, name, age):
        """
        Initializes the Person object.

        Args:
            name (str): The person's name.
            age (int): The person's age.
        """
        self.name = name
        self.age = age

    def display(self):
        """
        Displays the name and age of the person.

        Returns:
            None
        """
        print(f"Name: {self.name}, Age: {self.age}")


class Student(Person):
    """
    Student class inherits from Person and adds a school attribute.

    Attributes:
        name (str): The student's name.
        age (int): The student's age.
        school (str): The school the student attends.
    """

    def __init__(self, name, age, school):
        """
        Initializes the Student object.

        Args:
            name (str): The student's name.
            age (int): The student's age.
            school (str): The school the student attends.
        """
        super().__init__(name, age)
        self.school = school

    def displayStudent(self):
        """
        Displays the name, age, and school of the student.

        Returns:
            None
        """
        print(f"Name: {self.name}, Age: {self.age}, School: {self.school}")


#### test your class
if __name__ == "__main__":
    student = Student("Sara", 20, "Cairo University")
    student.displayStudent()
```
