## 1. Write a function that inputs a number and prints the multiplication table of that number

```python
def multiplicationTable(num):
    """Prints the multiplication table of num from 1 to 10."""
    for i in range(1, 11):
        print(f"{num} * {i} = {num * i}")
```

```python
multiplicationTable(9)
```

## 2. Write a program to print twin primes less than 1000
If two consecutive odd numbers are both prime then they are known as twin primes.

```python
def isPrime(n):
    """Returns True if n is a prime number, False otherwise."""
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True


def twinPrimes(limit):
    """Prints all twin prime pairs less than limit."""
    print("Twin Prime:")
    for n in range(2, limit - 2):
        if isPrime(n) and isPrime(n + 2):
            print(f"{n} and {n + 2}")
```

```python
twinPrimes(1000)
```

## 3. Write a program to find out the prime factors of a number
Example: prime factors of 56 - 2, 2, 2, 7

```python
def primeFactors(num):
    """Returns a list of prime factors of num."""
    factors = []
    divisor = 2
    while divisor * divisor <= num:
        while num % divisor == 0:
            factors.append(divisor)
            num //= divisor
        divisor += 1
    if num > 1:
        factors.append(num)
    return factors
```

```python
primeFactors(56)
```

## 4. Write a function that converts a decimal number to binary number

```python
def decToBin(num):
    """Converts a decimal number to its binary representation (as a string)."""
    if num == 0:
        return "0"
    binary_digits = []
    n = num
    while n > 0:
        binary_digits.append(str(n % 2))
        n //= 2
    return "".join(reversed(binary_digits))
```

```python
print(decToBin(11))
```

## 5. Print all perfect numbers in a given range
A number is called perfect if the sum of its proper divisors equals the number itself.
Example: 28 is perfect since 1 + 2 + 4 + 7 + 14 = 28.

```python
def properDivisorsSum(num):
    """Returns the sum of proper divisors of num."""
    if num == 0:
        return 0
    total = 0
    for i in range(1, num):
        if num % i == 0:
            total += i
    return total


def perfectNums(start, end):
    """Prints all perfect numbers in the inclusive range [start, end]."""
    for num in range(start, end + 1):
        if num == properDivisorsSum(num) and num != 0:
            print(num)
        elif num == 0:
            print(num)
```

```python
perfectNums(0, 100)
```
