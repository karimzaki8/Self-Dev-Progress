```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

n = int(input())
result = []

for i in range(n):
    result.append(fibonacci(i))

print(result)
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

n = int(input())
result = []

for i in range(n):
    result.append(fibonacci(i))

print(result)
Remove One Item from a Set
my_set = {1, 2, 3, 4, 5}

my_set.remove(3)

print(my_set)
Remove Multiple Items from a Set
my_set = {1, 2, 3, 4, 5}
items_to_remove = {1, 4}

my_set.difference_update(items_to_remove)

print(my_set)
Find the Index of an Item in a Tuple
my_tuple = ('apple', 'banana', 'orange', 'grape', 'banana')

print(my_tuple.index('orange'))
Convert a Sentence into a List of Words
sentence = input()

words = sentence.split()

print(words)
Palindrome Checker
text = input()

if text == text[::-1]:
    print("Palindrome")
else:
    print("Not a Palindrome")
Factorial Calculator
n = int(input())

factorial = 1

for i in range(1, n + 1):
    factorial *= i

print(factorial)
List of Squares (Using Loop)
squares = []

for i in range(1, 11):
    squares.append(i ** 2)

print(squares)
List of Squares (Using List Comprehension)
squares = [i ** 2 for i in range(1, 11)]

print(squares)
Sum of Even Numbers in a List
def sum_even(numbers):
    total = 0

    for num in numbers:
        if num % 2 == 0:
            total += num

    return total

numbers = [1, 2, 3, 4, 5, 6]

print(sum_even(numbers))
Reverse a List
def reverse_list(lst):
    return lst[::-1]

lst = [1, 2, 3]

print(reverse_list(lst))
Median of a List
def median(numbers):
    numbers.sort()
    n = len(numbers)

    if n % 2 == 1:
        return numbers[n // 2]
    else:
        return (numbers[n // 2 - 1] + numbers[n // 2]) / 2

numbers = [5, 2, 7, 3, 1]

print(median(numbers))
Mode of a List
def mode(lst):
    max_count = 0
    mode_value = None

    for num in lst:
        count = lst.count(num)
        if count > max_count:
            max_count = count
            mode_value = num

    return mode_value

lst = [5, 2, 7, 3, 2, 1, 5]

print(mode(lst))
Grade Calculator Based on Marks
mark = int(input())

if 85 <= mark <= 100:
    print("A")
elif 75 <= mark < 85:
    print("B")
elif 65 <= mark < 75:
    print("C")
elif 50 <= mark < 65:
    print("D")
else:
    print("F")
```
