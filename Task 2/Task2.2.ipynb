### 1. Add an item to a tuple

```python
my_tuple = (1, 2, 3)
item_to_add = 4

new_tuple = list(my_tuple)
new_tuple.append(item_to_add)
new_tuple = tuple(new_tuple)
print(new_tuple)
```

### 2. Sum all the items in a list

```python
numbers = [5, 10, 15, 20]

total = 0
for n in numbers:
    total += n
print(total)
```

### 3. Multiply all items in a list

```python
numbers = [2, 3, 4]

result = 1
for i in numbers:
    result *= i
print(result)
```

### 4. Get the smallest number from a list

```python
numbers = [10, 3, 25, 1, 7]

smallest = numbers[0]
for i in numbers:
    if i < smallest:
        smallest = i
print(smallest)
```

### 5. Get the largest number from a list

```python
numbers = [10, 3, 25, 1, 7]

largest = numbers[0]
for i in numbers:
    if i > largest:
        largest = i
print(largest)
```

### 6. Count number of strings with length ≥ 2

```python
words = ["hi", "a", "hello", "no", "on"]

count = 0
for i in words:
    if len(i) >= 2:
        count += 1
print(count)
```

### 7. Clone or copy a list

```python
original = [1, 2, 3, 4]

cloned = list(original)
print(cloned)
original[2] = 4
print(cloned)
print(original)
```

### 8. Remove item(s) from a set

```python
my_set = {1, 2, 3, 4, 5}

my_set.discard(2)
print(my_set)
```

### 9. Check if a set is subset of another

```python
set1 = {1, 2}
set2 = {1, 2, 3, 4}

print(set1.issubset(set2))
```

### 10. Remove all elements from a set

```python
my_set = {1, 2, 3}

my_set.clear()
print(my_set)
```

### 11. Find max and min values in a set

```python
my_set = {10, 3, 25, 1, 7}

current = my_set.pop()
largest = current
smallest = current

while my_set:
    current = my_set.pop()
    if current > largest:
        largest = current
    if current < smallest:
        smallest = current
print("Max:", largest)
print("Min:", smallest)
```

### 12. Find the index of an item in a tuple

```python
my_tuple = (10, 20, 30, 40)

index_value = my_tuple.index(30)
print(index_value)
```

### 13. Convert a tuple to a dictionary

```python
my_tuple = (("a", 1), ("b", 2), ("c", 3))

my_dict = dict(my_tuple)
print(my_dict)
```

### 14. Unzip a list of tuples into individual lists

```python
pairs = [(1, 2), (3, 4), (5, 6)]

list1, list2 = zip(*pairs)
list1 = list(list1)
list2 = list(list2)
print(list1)
print(list2)
```

### 15. Reverse a tuple

```python
my_tuple = (1, 2, 3, 4, 5)

reversed_tuple = my_tuple[::-1]
print(reversed_tuple)
```

### 16. Convert a list of tuples into a dictionary

```python
pairs = [("a", 1), ("b", 2), ("c", 3)]

my_dict = {key: value for key, value in pairs}
print(my_dict)
```

### 17. Replace the last value of each tuple in a list

```python
data = [(10, 20, 40), (40, 50, 60), (70, 80, 90)]

new_list = [i[:-1] + (100,) for i in data]
print(new_list)
```

### 18. Sort a tuple by its float element

```python
items = (("item1", 10.5), ("item2", 5.7), ("item3", 8.9))

sorted_items = sorted(items, key=lambda x: float(x[1]))
print(sorted_items)
```

## Challenge Part

#### Find the longest word in a list

```python
words = ["hi", "hello", "fantastic", "world"]

longest = words[0]
for i in words:
    if len(i) > len(longest):
        longest = i
print(longest)
```

#### Merge two dictionaries

```python
d1 = {"a": 1, "b": 2}
d2 = {"c": 3, "d": 4}

result = d1 | d2
print(result)
```

#### Sum values in a dictionary

```python
data = {"a": 10, "b": 20, "c": 30}

total = 0
for value in data.values():
    total += value
print(total)
```

#### New Challenge – Sum all nested scores

```python
scores = {
    "student1": {"math": 80, "science": 90},
    "student2": {"math": 75, "science": 85},
    "student3": {"math": 92, "science": 88},
}

total = 0
for student_data in scores.values():
    for score in student_data.values():
        total += score
print(total)
```
